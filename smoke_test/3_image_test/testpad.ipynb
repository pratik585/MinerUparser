```python
from mineru.parser import MinerUParser
from rich import print as rprint
from mineru.parser import ParseResult
```

```python
parser = MinerUParser(
    tier="standard",
    parse_mode="auto",
    image_analysis=True,
    vlm_config=None
)
```

```python
import logging
help(logging)
```

```python
from pathlib import Path
from mineru.parser.writer import FileBasedDataWriter

image_dir = Path("3images")
image_paths = [p for p in sorted(image_dir.iterdir()) if p.is_file()]
output_base_dir = Path("prediction_3_from_test")

failed_to_parse = []

for img_path in image_paths:
    file_name = img_path.stem
    output_dir = output_base_dir/file_name
    
    if output_dir.is_dir():
        continue
    else:
        try:   
            result = parser.parse(img_path)
        except Exception as e:
            failed_to_parse.append(img_path)
            
        else: 
            writer = FileBasedDataWriter(str(Path("prediction_3_from_test")/file_name))
            result.save(writer)
```
